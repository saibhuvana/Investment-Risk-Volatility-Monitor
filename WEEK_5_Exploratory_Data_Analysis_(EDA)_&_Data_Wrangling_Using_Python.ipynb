```python
import pandas as pd
import numpy as np
```

```python
xls = pd.ExcelFile("finanace_Dashboard.xlsx")
xls.sheet_names
```

```python
df = pd.read_excel(
    "finanace_Dashboard.xlsx",
    sheet_name="finance_Cleaned"
)
```

```python
df.columns
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include='number').columns
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.columns
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df_2023 = df[df['year'] == 2023]
```

```python
df.loc[0:5]
```

```python
df.iloc[0:5, 0:4]
```

```python
df['price_change'] = df['close'] - df['open']
```

```python
df['trend'] = df['price_change'].apply(
    lambda x: 'Profit' if x > 0 else 'Loss'
)
```

```python
df.groupby('name')['close'].mean()
```

```python
df.groupby('name')['volume'].sum().sort_values(ascending=False)
```

```python
pd.pivot_table(
    df,
    values='close',
    index='name',
    columns='trend',
    aggfunc='mean'
)
```

```python
jan = df[df['month'] == 1]
feb = df[df['month'] == 2]
merged = pd.concat([jan, feb])
merged.head()
```
